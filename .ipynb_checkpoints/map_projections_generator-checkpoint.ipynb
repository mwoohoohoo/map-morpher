```python
from pathlib import Path
import subprocess


# ============================================================
# CONFIGURATION
# ============================================================

# Project folder = folder containing this notebook/script
BASE_DIR = Path.cwd()

# Existing Natural Earth Shapefile
INPUT_FILE = (
    BASE_DIR
    / "data"
    / "ne_50m_admin_0_countries"
    / "ne_50m_admin_0_countries.shp"
)

# Output directory
OUTPUT_DIR = (
    BASE_DIR
    / "data"
    / "projected_world"
)


# ============================================================
# PROJECTIONS
# ============================================================

PROJECTIONS = {
    "mercator": "+proj=merc",
    "robinson": "+proj=robin",
    "dymaxion": "+proj=dymaxion",
    "gall_peters": "+proj=cea +lat_ts=45",
    "goode_homolosine": "+proj=igh",
    "narukawa2022_authagraph": "+proj=narukawa2022",
    "peirce_quincuncial": "+proj=peirce_q",
    "winkel_tripel": "+proj=wintri",
}


# ============================================================
# CHECK INPUT
# ============================================================

if not INPUT_FILE.exists():
    raise FileNotFoundError(
        f"Natural Earth Shapefile not found:\n\n"
        f"{INPUT_FILE}\n\n"
        f"Current working directory:\n"
        f"{BASE_DIR}"
    )


# ============================================================
# CHECK MAPSHAPER
# ============================================================

try:
    result = subprocess.run(
        ["npx", "mapshaper", "-v"],
        capture_output=True,
        text=True,
        check=True
    )

    print(f"Mapshaper: {result.stdout.strip()}")

except FileNotFoundError:
    raise RuntimeError(
        "npx was not found. Make sure Node.js/npm is installed."
    )

except subprocess.CalledProcessError as e:
    raise RuntimeError(
        "Mapshaper could not be run.\n\n"
        f"STDOUT:\n{e.stdout}\n\n"
        f"STDERR:\n{e.stderr}"
    )


# ============================================================
# CREATE OUTPUT DIRECTORY
# ============================================================

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# DISPLAY CONFIGURATION
# ============================================================

print("\n" + "=" * 60)
print("WORLD PROJECTION GENERATOR")
print("=" * 60)

print("\nInput:")
print(f"  {INPUT_FILE}")

print("\nOutput:")
print(f"  {OUTPUT_DIR}")

print("\nProjections:")

for name, projection in PROJECTIONS.items():
    print(f"  {name}: {projection}")


# ============================================================
# PROJECT EACH DATASET
# ============================================================

successful = []
failed = []
outline_successful = []
outline_failed = []


for name, projection in PROJECTIONS.items():

    # --------------------------------------------------------
    # Create projected countries
    # --------------------------------------------------------

    output_file = (
        OUTPUT_DIR
        / f"world_{name}.geojson"
    )

    print("\n" + "=" * 60)
    print(f"Creating: {output_file.name}")
    print(f"Projection: {projection}")
    print("=" * 60)

    command = [
        "npx",
        "mapshaper",
        str(INPUT_FILE),
        "-proj",
        projection,
        "-o",
        str(output_file),
    ]

    try:

        result = subprocess.run(
            command,
            capture_output=True,
            text=True,
            check=True
        )

        if result.stdout:
            print(result.stdout)

        print(f"✓ Created: {output_file.name}")

        successful.append(name)

    except subprocess.CalledProcessError as e:

        print(f"✗ FAILED: {output_file.name}")

        if e.stdout:
            print("\nSTDOUT:")
            print(e.stdout)

        if e.stderr:
            print("\nSTDERR:")
            print(e.stderr)

        failed.append(name)


    # --------------------------------------------------------
    # Create projection outline / neatline
    # --------------------------------------------------------

    outline_file = (
        OUTPUT_DIR
        / f"outline_{name}.geojson"
    )

    print(f"\nCreating outline: {outline_file.name}")

    outline_command = [
        "npx",
        "mapshaper",
        str(INPUT_FILE),
        "-proj",
        projection,
        "densify",
        "-graticule",
        "outline",
        "name=neatline",
        "-o",
        "target=neatline",
        str(outline_file),
    ]

    try:

        result = subprocess.run(
            outline_command,
            capture_output=True,
            text=True,
            check=True
        )

        if result.stdout:
            print(result.stdout)

        print(f"✓ Created: {outline_file.name}")

        outline_successful.append(name)

    except subprocess.CalledProcessError as e:

        print(f"✗ FAILED: {outline_file.name}")

        if e.stdout:
            print("\nSTDOUT:")
            print(e.stdout)

        if e.stderr:
            print("\nSTDERR:")
            print(e.stderr)

        outline_failed.append(name)


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("SUMMARY")
print("=" * 60)

print(
    f"\nCountry files created: "
    f"{len(successful)} / {len(PROJECTIONS)}"
)

for name in successful:
    print(f"  ✓ world_{name}.geojson")


if failed:

    print("\nCountry files that failed:")

    for name in failed:
        print(f"  ✗ world_{name}.geojson")


print(
    f"\nProjection outlines created: "
    f"{len(outline_successful)} / {len(PROJECTIONS)}"
)

for name in outline_successful:
    print(f"  ✓ outline_{name}.geojson")


if outline_failed:

    print("\nProjection outlines that failed:")

    for name in outline_failed:
        print(f"  ✗ outline_{name}.geojson")


print("\nOutput directory:")
print(f"  {OUTPUT_DIR.resolve()}")
```

```python
from pathlib import Path
import subprocess


# ============================================================
# CONFIGURATION
# ============================================================

BASE_DIR = Path.cwd()

INPUT_FILE = (
    BASE_DIR
    / "data"
    / "ne_50m_admin_0_countries"
    / "ne_50m_admin_0_countries.shp"
)

OUTPUT_DIR = (
    BASE_DIR
    / "data"
    / "projected_world"
)

WORLD_OUTPUT = (
    OUTPUT_DIR
    / "world_mercator.geojson"
)

OUTLINE_OUTPUT = (
    OUTPUT_DIR
    / "outline_mercator.geojson"
)

# Web Mercator latitude limit
LAT_LIMIT = 85.05112878


# ============================================================
# CHECK INPUT
# ============================================================

if not INPUT_FILE.exists():
    raise FileNotFoundError(
        f"Natural Earth Shapefile not found:\n\n"
        f"{INPUT_FILE}\n\n"
        f"Current working directory:\n"
        f"{BASE_DIR}"
    )


# ============================================================
# CHECK MAPSHAPER
# ============================================================

try:
    result = subprocess.run(
        ["npx", "mapshaper", "-v"],
        capture_output=True,
        text=True,
        check=True
    )

    print(f"Mapshaper: {result.stdout.strip()}")

except FileNotFoundError:
    raise RuntimeError(
        "npx was not found. Make sure Node.js/npm is installed."
    )

except subprocess.CalledProcessError as e:
    raise RuntimeError(
        "Mapshaper could not be run.\n\n"
        f"STDOUT:\n{e.stdout}\n\n"
        f"STDERR:\n{e.stderr}"
    )


# ============================================================
# CREATE OUTPUT DIRECTORY
# ============================================================

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# COMMON CLIP BBOX
# ============================================================

BBOX = (
    f"-180,-{LAT_LIMIT},"
    f"180,{LAT_LIMIT}"
)


# ============================================================
# CREATE MERCATOR WORLD
# ============================================================

print("\n" + "=" * 60)
print("CREATING MERCATOR WORLD")
print("=" * 60)

print(f"\nLatitude limit: ±{LAT_LIMIT}°")
print(f"Input:  {INPUT_FILE}")
print(f"Output: {WORLD_OUTPUT}")


world_command = [
    "npx",
    "mapshaper",
    str(INPUT_FILE),

    # Clip BEFORE projection
    "-clip",
    f"bbox={BBOX}",

    # Project
    "-proj",
    "+proj=merc",

    # Output
    "-o",
    str(WORLD_OUTPUT),
]


try:

    result = subprocess.run(
        world_command,
        capture_output=True,
        text=True,
        check=True
    )

    if result.stdout:
        print(result.stdout)

    print(
        f"\n✓ Created: "
        f"{WORLD_OUTPUT.name}"
    )

except subprocess.CalledProcessError as e:

    print("\n✗ Failed to create Mercator world")

    if e.stdout:
        print("\nSTDOUT:")
        print(e.stdout)

    if e.stderr:
        print("\nSTDERR:")
        print(e.stderr)

    raise


# ============================================================
# CREATE MERCATOR OUTLINE
# ============================================================

print("\n" + "=" * 60)
print("CREATING MERCATOR OUTLINE")
print("=" * 60)

print(f"\nLongitude: -180° to 180°")
print(f"Latitude:  -{LAT_LIMIT}° to {LAT_LIMIT}°")
print(f"Output:    {OUTLINE_OUTPUT}")


outline_command = [
    "npx",
    "mapshaper",

    # Create a geographic rectangle corresponding
    # exactly to the clipped Mercator domain
    "-rectangle",
    f"bbox={BBOX}",

    # Name the layer
    "name=neatline",

    # Project the rectangle
    "-proj",
    "+proj=merc",

    # Write only the neatline layer
    "-o",
    "target=neatline",
    str(OUTLINE_OUTPUT),
]


try:

    result = subprocess.run(
        outline_command,
        capture_output=True,
        text=True,
        check=True
    )

    if result.stdout:
        print(result.stdout)

    print(
        f"\n✓ Created: "
        f"{OUTLINE_OUTPUT.name}"
    )

except subprocess.CalledProcessError as e:

    print("\n✗ Failed to create Mercator outline")

    if e.stdout:
        print("\nSTDOUT:")
        print(e.stdout)

    if e.stderr:
        print("\nSTDERR:")
        print(e.stderr)

    raise


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("DONE")
print("=" * 60)

print("\nGenerated files:")

print(f"  ✓ {WORLD_OUTPUT}")
print(f"  ✓ {OUTLINE_OUTPUT}")

print(
    "\nMercator is clipped to "
    f"±{LAT_LIMIT}° before projection."
)

print(
    "\nAntarctica is retained, but the geometry "
    "south of the Web Mercator cutoff is removed."
)
```
