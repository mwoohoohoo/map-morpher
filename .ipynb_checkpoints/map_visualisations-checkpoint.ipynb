```python
import geopandas as gpd
import matplotlib.pyplot as plt


def plot_projection(
    world_filepath,
    outline_filepath=None,
    title=None,
    width=10
):

    # --------------------------------------------------------
    # Load projected world
    # --------------------------------------------------------

    world = gpd.read_file(world_filepath)

    # --------------------------------------------------------
    # Determine natural aspect ratio
    # --------------------------------------------------------

    xmin, ymin, xmax, ymax = world.total_bounds

    map_width = xmax - xmin
    map_height = ymax - ymin

    height = width * (map_height / map_width)

    # --------------------------------------------------------
    # Create figure
    # --------------------------------------------------------

    fig, ax = plt.subplots(
        figsize=(width, height)
    )

    # --------------------------------------------------------
    # Plot countries
    # --------------------------------------------------------

    world.plot(
        ax=ax,
        facecolor="lightgrey",
        edgecolor="black",
        linewidth=0.4,
        aspect="equal"
    )

    # --------------------------------------------------------
    # Plot projection outline
    # --------------------------------------------------------

    if outline_filepath is not None:

        outline = gpd.read_file(
            outline_filepath
        )

        outline.plot(
            ax=ax,
            facecolor="none",
            edgecolor="black",
            linewidth=1.5,
            aspect="equal"
        )

    # --------------------------------------------------------
    # Ensure equal x/y scaling
    # --------------------------------------------------------

    ax.set_aspect(
        "equal",
        adjustable="box"
    )

    ax.set_axis_off()

    if title:
        ax.set_title(title)

    plt.tight_layout()

    return fig, ax
```

```python
fig, ax = plot_projection(
    "data/projected_world/world_dymaxion.geojson",
    "data/projected_world/outline_dymaxion.geojson",
    title="Dymaxion"
)

plt.show()
```
